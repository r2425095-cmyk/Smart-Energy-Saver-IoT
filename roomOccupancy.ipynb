```python
import serial
import time

#######################
# import joblib
# import cv2
# import numpy as np

# # 1. Load the model
# model = joblib.load('occupancy_model.pkl')

# def predict_frame(frame):
#     # Process image exactly like training (Grayscale, 64x64)
#     gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
#     resized = cv2.resize(gray, (64, 64))
    
#     # Flatten to 1D array
#     flattened = resized.flatten().reshape(1, -1)
    
#     # Scale if necessary
#     # flattened = scaler.transform(flattened)

#     # Predict
#     prediction = model.predict(flattened)
#     return "Occupied" if prediction[0] == 1 else "Empty"
########################

# class_names = ['Empty', 'Occupied']
# # 2. Setup Serial
# ser = serial.Serial('COM4', 9600)
# time.sleep(2)

# cap = cv2.VideoCapture(0)

# while True:
#     ret, frame = cap.read()
#     if not ret: break

#     # 3. Pre-process the image for the model
#     # (Typical models expect 224x224 or 150x150)
#     img = cv2.resize(frame, (100, 100))
#     img = img.astype('float32') / 255.0
#     img = np.expand_dims(img, axis=0)

#     # 4. Make Prediction
#     prediction = model.predict(img)
#     index = np.argmax(prediction)
#     status = class_names[index]

#     # 5. Send command to Arduino
#     if status == 'Occupied':
#         ser.write(b'1')
#         label_color = (0, 255, 0)
#     else:
#         ser.write(b'0')
#         label_color = (0, 0, 255)

#     # Visual Feedback
#     cv2.putText(frame, f"Room: {status}", (20, 40), 
#                 cv2.FONT_HERSHEY_SIMPLEX, 1, label_color, 2)
#     cv2.imshow("Smart Energy System", frame)

#     if cv2.waitKey(1) & 0xFF == ord('q'):
#         break

# cap.release()
# cv2.destroyAllWindows()
# ser.close()



import streamlit as st
import os
import time
import joblib
from skimage import io, transform

# --- APP CONFIGURATION ---
st.set_page_config(page_title="Smart Energy Simulator", layout="centered")
st.title("💡 Smart Room Energy Management")
st.write("Simulating room occupancy detection to control fan and lights.")

# Placeholders for dynamic content
image_placeholder = st.empty()
status_placeholder = st.empty()
timer_placeholder = st.empty()

# --- SETUP SERIAL & MODEL ---
MODEL_PATH = "occupancy_model.pkl"
IMAGE_FOLDER = "Rooms"

@st.cache_resource
def load_resources():
    model = joblib.load(MODEL_PATH)
    # Update 'COM3' to your Arduino's port
    try:
        ser = serial.Serial('COM4', 9600, timeout=1)
        time.sleep(2)
        return model, ser
    except:
        return model, None

model, ser = load_resources()
if ser is None:
    st.warning("⚠️ Arduino not detected. Running in software-only mode.")

# --- SIMULATION LOOP ---
images = [f for f in os.listdir(IMAGE_FOLDER) if f.endswith(('.png', '.jpg', '.jpeg'))]

while True:
    for img_name in images:
        img_path = os.path.join(IMAGE_FOLDER, img_name)
        
        # 1. Load and Process with scikit-image
        img = io.imread(img_path)
        img_resized = transform.resize(img, (64, 64)) # Match your training size
        img_flattened = img_resized.flatten().reshape(1, -1)

        # 2. Predict
        prediction = model.predict(img_flattened)
        is_occupied = prediction[0] == 1
        status = "Occupied" if is_occupied else "Empty"
        
        # 3. Update Streamlit UI
        image_placeholder.image(img, caption=f"Processing: {img_name}", use_container_width=True)
        
        color = "green" if is_occupied else "red"
        status_placeholder.markdown(f"### Current State: :{color}[{status}]")

        # 4. Signal Arduino
        if ser:
            ser.write(b'1' if is_occupied else b'0')

        # 5. Timer Logic (1 Minute per image)
        for remaining in range(60, 0, -1):
            timer_placeholder.write(f"Next image in {remaining} seconds...")
            time.sleep(1)
```
